```python
# Paradoja de las pruebas diagnósticas
# Sensibilidad = 99%, Especificidad = 99%

def calcular_ppv(sensibilidad, especificidad, prevalencia):
    """
    Calcula el Valor Predictivo Positivo (PPV) usando Bayes
    """
    numerador = sensibilidad * prevalencia
    denominador = (sensibilidad * prevalencia) + \
                  ((1 - especificidad) * (1 - prevalencia))
    return numerador / denominador


def simulacion(poblacion, sensibilidad, especificidad, prevalencia):
    """
    Convierte probabilidades en números reales
    """
    casos_reales = int(poblacion * prevalencia)
    sanos = poblacion - casos_reales

    verdaderos_positivos = int(casos_reales * sensibilidad)
    falsos_positivos = int(sanos * (1 - especificidad))

    ppv = calcular_ppv(sensibilidad, especificidad, prevalencia)

    print(f"Población total: {poblacion}")
    print(f"Casos reales: {casos_reales}")
    print(f"Verdaderos positivos: {verdaderos_positivos}")
    print(f"Falsos positivos: {falsos_positivos}")
    print(f"PPV ≈ {ppv*100:.2f}%")

# Parámetros
poblacion = 10000
sensibilidad = 0.99
especificidad = 0.99
prevalencia = 0.001   # 0.1%

simulacion(poblacion, sensibilidad, especificidad, prevalencia)
```
